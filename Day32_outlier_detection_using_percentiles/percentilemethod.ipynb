## **Outlier_detection_using_percentiles**

#### **Import libraries**

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('weight-height.csv')
```

```python
upper_limit = df['Height'].quantile(0.99)
upper_limit
```

```python
lower_limit = df['Height'].quantile(0.01)
lower_limit
```

```python
new_df = df[(df['Height'] <= 74.78) & (df['Height'] >= 58.13)]
```

```python
new_df['Height'].describe()
```

```python
df['Height'].describe()
```

```python
sns.histplot(new_df['Height'],kde=True)
```

```python
sns.boxplot(new_df['Height'])
```

## **Capping --> Winsorization**


```python
df['Height'] = np.where(df['Height'] >= upper_limit,
        upper_limit,
        np.where(df['Height'] <= lower_limit,
        lower_limit,
        df['Height']))
```

```python
df.shape
```

```python
df['Height'].describe()
```

```python
sns.histplot(df['Height'], kde=True, color='red', label='After Capping')
```

```python
sns.boxplot(df['Height'], color='red', label='After Capping')
```
