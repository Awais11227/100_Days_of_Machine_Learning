## **Elastic Net Regression in Machine Learning**

- Today, I explored Elastic Net Regression, a powerful regularization technique that combines the strengths of both Ridge Regression and Lasso Regression.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

### **LOad Dataset**

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

## **Apply Ridge Regression**

```python
# Ridge 
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

## **Apply Lasso Regression**

```python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

## **Apply ElasticNet Regression**

```python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

# **Compare**


```python
results = {
    "Linear Regression": 0.4399,
    "Ridge Regression": 0.4520,
    "Lasso Regression": 0.4411,
    "Elastic Net": 0.4531
}

for model, score in results.items():
    print(model, ":", score)
```

# `Model`	`R² Score`
- Linear Regression	0.4399
- Ridge Regression	0.4520
- Lasso Regression	0.4411
- Elastic Net Regression	0.4531

## **Result**

Elastic Net achieved the highest R² Score among all tested models.
 - This shows that combining L1 + L2 regularization can outperform using either one alone.
