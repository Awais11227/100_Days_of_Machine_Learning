```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

## Generate Dataset

```python
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=30
)

print(X.shape)
print(y.shape)
```

```python
# Visualize the data
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.title("Binary Classification Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

## Step Activation Function

```python
def step(z):
    return 1 if z > 0 else 0
```

```python
def perceptron(X, y):

    X = np.insert(X, 0, 1, axis=1)

    weights = np.ones(X.shape[1])

    lr = 0.1

    for i in range(1000):

        j = np.random.randint(0,100)

        y_hat = step(np.dot(X[j], weights))

        weights = weights + lr * (y[j] - y_hat) * X[j]

    return weights[0], weights[1:]
```

## Train 

```python
intercept_, coef_ = perceptron(X, y)

print("Intercept:", intercept_)
print("Coefficients:", coef_)
```

## Perceptron Decision Boundary

```python
m = -(coef_[0] / coef_[1])
b = -(intercept_ / coef_[1])

x_input = np.linspace(-3,3,100)
y_input = m * x_input + b

plt.figure(figsize=(10,6))
plt.plot(x_input, y_input, color='red', linewidth=3, label='Perceptron')
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)

plt.ylim(-3,2)
plt.legend()
plt.show()
```

## logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(X,y)
```

```python
m1 = -(lor.coef_[0][0] / lor.coef_[0][1])
b1 = -(lor.intercept_[0] / lor.coef_[0][1])

x_input1 = np.linspace(-3,3,100)
y_input1 = m1 * x_input1 + b1
```

## **Compare Both**

```python
plt.figure(figsize=(10,6))

plt.plot(x_input, y_input, color='red', linewidth=3, label='Perceptron')
plt.plot(x_input1, y_input1, color='black', linewidth=3, label='Logistic Regression')

plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)

plt.ylim(-3,2)
plt.legend()
plt.title("Perceptron vs Logistic Regression")
plt.show()
```

## Sigmoid Function

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

### Modified Perceptron

```python
def perceptron_sigmoid(X, y):

    X = np.insert(X, 0, 1, axis=1)

    weights = np.ones(X.shape[1])

    lr = 0.1

    for i in range(1000):

        j = np.random.randint(0,100)

        y_hat = sigmoid(np.dot(X[j], weights))

        weights = weights + lr * (y[j] - y_hat) * X[j]

    return weights[0], weights[1:]
```

```python
intercept2, coef2 = perceptron_sigmoid(X, y)

m2 = -(coef2[0] / coef2[1])
b2 = -(intercept2 / coef2[1])

x_input2 = np.linspace(-3,3,100)
y_input2 = m2 * x_input2 + b2
```

## Compare All

```python
plt.figure(figsize=(10,6))

plt.plot(x_input, y_input, color='red', linewidth=3, label='Perceptron')
plt.plot(x_input1, y_input1, color='black', linewidth=3, label='Logistic Regression')
plt.plot(x_input2, y_input2, color='brown', linewidth=3, label='Sigmoid Perceptron')

plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)

plt.ylim(-3,2)
plt.legend()
plt.title("All Decision Boundaries")
plt.show()
```

## Animation Data

```python
def perceptron_animation(X, y):

    m = []
    b = []

    X = np.insert(X, 0, 1, axis=1)

    weights = np.ones(X.shape[1])

    lr = 0.1

    for i in range(200):

        j = np.random.randint(0,100)

        y_hat = step(np.dot(X[j], weights))

        weights = weights + lr * (y[j] - y_hat) * X[j]

        m.append(-(weights[1] / weights[2]))
        b.append(-(weights[0] / weights[2]))

    return m, b

m_list, b_list = perceptron_animation(X, y)
```

```python
fig, ax = plt.subplots(figsize=(9,5))

x_i = np.arange(-3,3,0.1)

ax.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)

line, = ax.plot(x_i, x_i*m_list[0] + b_list[0], 'r-', linewidth=2)

plt.ylim(-3,3)

def update(i):
    line.set_ydata(x_i*m_list[i] + b_list[i])
    ax.set_xlabel("Epoch " + str(i+1))
    return line,

anim = FuncAnimation(fig, update, frames=200, interval=100)

HTML(anim.to_jshtml())
```

### **Perceptron**

- Oldest binary classifier
- Uses step activation
- Works on linearly separable data

### **Logistic Regression**

- Uses sigmoid function
- Gives probabilities
- More stable than perceptron
