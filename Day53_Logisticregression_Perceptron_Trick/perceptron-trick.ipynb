## **LR - Perceptron Algorithm Trick**

Perceptron is one of the earliest machine learning algorithms used for binary classification.
It learns a linear decision boundary to separate two classes.

## **Import Libraries**

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
%matplotlib inline
import matplotlib.pyplot as plt
```

## **Generate Data**

```python
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=10
)

print(X.shape)
print(y.shape)
```

### **Visualize Dataset**

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.title("Generated Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
def step(z):
    return 1 if z > 0 else 0
```

## **Perceptron Code**

```python
def perceptron(X, y):

    X = np.insert(X, 0, 1, axis=1)

    weights = np.ones(X.shape[1])

    lr = 0.1

    for i in range(1000):

        j = np.random.randint(0, 100)

        y_hat = step(np.dot(X[j], weights))

        weights = weights + lr * (y[j] - y_hat) * X[j]

    return weights[0], weights[1:]
```

### **Train Model**

```python
intercept_, coef_ = perceptron(X, y)

print("Intercept:", intercept_)
print("Coefficients:", coef_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

## **Decision Boundary**

```python
m = -(coef_[0] / coef_[1])
b = -(intercept_ / coef_[1])

x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + b

plt.figure(figsize=(10,6))
plt.plot(x_input, y_input, color='red', linewidth=3)
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)

plt.ylim(-3,2)
plt.title("Perceptron Decision Boundary")
plt.show()
```

## Compare with Logistic Regression

```python
lor = LogisticRegression()

lor.fit(X, y)
```

### **Logistic Boundary**

```python
m1 = -(lor.coef_[0][0] / lor.coef_[0][1])
b1 = -(lor.intercept_ / lor.coef_[0][1])

x_input1 = np.linspace(-3, 3, 100)
y_input1 = m1 * x_input1 + b1
```

```python
plt.figure(figsize=(10,6))

plt.plot(x_input, y_input, color='red', linewidth=3, label="Perceptron")
plt.plot(x_input1, y_input1, color='black', linewidth=3, label="Logistic Regression")

plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)

plt.ylim(-3,2)

plt.legend()
plt.title("Perceptron vs Logistic Regression")
plt.show()
```
