```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

import ipywidgets as widgets
from ipywidgets import interact


def run_model(dataset='Binary', C=1.0, max_iter=100):

    # Dataset Selection
    if dataset == 'Binary':
        X, y = make_blobs(
            n_features=2,
            centers=2,
            random_state=6
        )
    else:
        X, y = make_blobs(
            n_features=2,
            centers=3,
            random_state=2
        )

    # Train Test Split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, random_state=42
    )

    # Model
    clf = LogisticRegression(
        C=C,
        max_iter=max_iter
    )

    clf.fit(X_train, y_train)

    # Accuracy
    y_pred = clf.predict(X_test)
    acc = accuracy_score(y_test, y_pred)

    # Meshgrid
    a = np.arange(X[:,0].min()-1, X[:,0].max()+1, 0.01)
    b = np.arange(X[:,1].min()-1, X[:,1].max()+1, 0.01)

    XX, YY = np.meshgrid(a, b)

    input_array = np.array(
        [XX.ravel(), YY.ravel()]
    ).T

    labels = clf.predict(input_array)

    # Plot
    plt.figure(figsize=(10,7))

    plt.contourf(
        XX, YY,
        labels.reshape(XX.shape),
        alpha=0.4,
        cmap='rainbow'
    )

    plt.scatter(
        X[:,0], X[:,1],
        c=y,
        cmap='rainbow',
        edgecolors='black'
    )

    plt.title(f'Accuracy = {round(acc,2)}')
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")

    plt.show()


interact(
    run_model,

    dataset=widgets.Dropdown(
        options=['Binary', 'Multiclass'],
        value='Binary',
        description='Dataset:'
    ),

    C=widgets.FloatSlider(
        value=1.0,
        min=0.01,
        max=10.0,
        step=0.1,
        description='C:'
    ),

    max_iter=widgets.IntSlider(
        value=100,
        min=50,
        max=1000,
        step=50,
        description='Iterations:'
    )
)
```
